#### Exercise 2

The `tickers` and `earnings` lists below contain tickers and corresponding earnings per share data for each ticker.

1. Create a dictionary called `eps_dict` with the tickers as keys and the earnings per share numbers as individual values for each key.

```python
# 1

tickers = ['AAPL','F','AMZN']
earnings = [1.50,2.7,1.2]

eps_dict = {}

for t,ticker in enumerate(tickers):
    eps = earnings[t]
    eps_dict[ticker] = eps

print(eps_dict)
```

```python
data = open('avocados.txt').readlines()
```

1. Using the `data` list, write code to answer the following question: If you purchased 3 avocados on each day that we have pricing data in 2017, how much money would you have spent on avocados in 2017?

    Hint: Recall, you can use the `in` statement to check whether some text is contained within a string. For example, if you want to check if `/2017` is included in your string, you could use the following code:

        if '/2017' in line:
            perform_some_task

```python
# 1

total_cost_2017 = 0

for i,line in enumerate(data):
    if i > 0:
        if '/2017' in line:
            price = float(line.split('|')[1])
            daily_cost = price*3
            total_cost_2017 = total_cost_2017 + daily_cost
            
print(f'The total cost of purchasing 3 avocados each day on all dates in 2017 for which we have price data would be ${total_cost_2017:.2f}.')
```

```python
# 1 Alternative solution

cost_2017 = []

for i,line in enumerate(data):
    if i > 0:
        if '/2017' in line:
            price = float(line.split('|')[1])
            daily_cost = price*3
            cost_2017.append(daily_cost)
            
total_cost_2017 = sum(cost_2017)

print(f'The total cost of purchasing 3 avocados each day on all dates in 2017 for which we have price data would be ${total_cost_2017:.2f}.')
```

2. How many days have pricing data in 2017 in our dataset?

```python
# 2

count_2017 = 0

for i,line in enumerate(data):
    if i > 0:
        if '/2017' in line:
            count_2017 = count_2017 + 1
            
print(f'We have {count_2017} days of pricing data in 2017.')
```

```python
# 2 Alternative (assuming you used # 1 Alternative solution)

count_2017 = len(cost_2017)

print(f'We have {count_2017} days of pricing data in 2017.')
```
