# Module 3 — pandas: load, calculate, filter


`fin400-survey.csv` is a sample of 900 adults: `seqn` (an id), `age`, `gender`,
`weight` (kg), `height` (cm), `waist` (cm) and `hincome` (`high` when household income is
over $120,000, otherwise `not`).

A DataFrame is a table with named columns. Almost every exercise starts the same way:
read the file, add a column, look at the numbers, then keep the rows you care about.


## 1. Read it and look at it


```python
import pandas as pd

data = pd.read_csv("fin400-survey.csv")

print(data.shape)
print(list(data.columns))
print(data.head())
```

```python
print(data.dtypes)
print(data["age"].mean())
print(data["weight"].max())
```

## 2. A new column is arithmetic on columns


```python
data["bmi"] = data["weight"] / (data["height"] / 100) ** 2

print(data[["weight", "height", "bmi"]].head())
```

The parentheses matter: `height / 100` converts centimetres to metres, and that whole
quantity is squared. `data["bmi"]` is then an ordinary column you can summarise.


## 3. Summaries


```python
print(data["bmi"].describe())
print(data["bmi"].quantile([0.1, 0.9]))
```

```python
print(data["bmi"].median())
print(data["bmi"].std())
```

`describe()` gives count, mean, spread and quartiles in one object. `quantile` cuts the
distribution where you ask — the 10th and 90th percentile leave 10% in each tail.


## 4. Logical indexing


```python
men = data[data["gender"] == "male"]
print(len(men))

young_men = data[(data["gender"] == "male") & (data["age"] <= 30)]
print(len(young_men))

tall_young_men = data[(data["gender"] == "male") & (data["age"] <= 30) & (data["height"] > 182.9)]
print(len(tall_young_men))
```

A condition on a column produces a column of `True`/`False`, and `data[...]` keeps the
rows where it is `True`. Join conditions with `&` and keep each one in parentheses — the
word `and` does not work on a whole column.


## 5. Counting categories and sorting


```python
print(data["gender"].value_counts())
print(data["hincome"].value_counts())

print(data.sort_values("bmi", ascending=False)[["age", "gender", "bmi"]].head())
```
