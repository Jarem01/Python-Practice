# Module 2 — Functions, scope and the statistics library


A function is a name for a calculation you want to reuse. This module covers the three
things that trip people up — what a function returns, which names it can see, and which
arguments are optional — and then the `statistics` library the class uses for the normal
distribution.


## 1. A function returns a value


```python
def average(values):
    total = 0
    count = 0
    for value in values:
        total += value
        count += 1
    return total / count


print(average([0.01, 0.02, 0.03]))
print(average([4, 8]))
```

`return` hands the answer back to whoever called the function. A function that prints
instead of returning cannot be reused inside another calculation, which is why the
exercises ask for `return`.


## 2. Scope: the name inside is not the name outside


```python
rate = 0.05


def grow(balance):
    rate = 0.10          # a new, local name
    return balance * (1 + rate)


print(grow(1000))
print(rate)
```

Without `return`, the work is thrown away — and the outer `rate` is untouched either
way. If a function needs to change something, it has to return it.


```python
def grow(balance):
    rate = 0.10
    return balance * (1 + rate)


rate = grow(1000)
print(rate)
```

## 3. Optional arguments


```python
import statistics


def draw(n, mu=0, sigma=1):
    return statistics.NormalDist(mu, sigma).samples(n)


print(len(draw(5)))
print(round(statistics.mean(draw(2000, 10, 2)), 2))
```

Calling with keywords — `draw(2000, mu=10, sigma=2)` — says out loud which number is
which. The call above relies on position instead.


## 4. The normal CDF


```python
import statistics as s

dist = s.NormalDist(100, 15)
print(round(dist.cdf(100), 4))
print(round(1 - dist.cdf(130), 4))
```

`cdf(x)` is the probability of a value at or below `x`. The second line is the upper
tail — the probability of being above 130 — which is why it subtracts from 1.


```python
def cdf_values(values, mu=0, sigma=1):
    dist = s.NormalDist(mu, sigma)
    out = []
    for value in values:
        out.append(dist.cdf(value))
    return out


print([round(v, 4) for v in cdf_values([85, 100, 115], 100, 15)])
```
