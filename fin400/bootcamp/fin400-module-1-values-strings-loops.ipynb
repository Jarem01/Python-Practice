## 1. Numbers and strings


```python
price = 84.5
shares = 120
position = price * shares

print(position)
print(round(position, 2))
```

```python
ticker = "brk.b"

print(ticker.upper())
print(ticker.replace(".", "-"))
print(len(ticker))
```

An f-string puts a value straight into text, and the part after the colon controls the
format: `,` groups thousands, `.2f` keeps two decimals.


```python
name = "Berkshire"
print(f"{name} is {len(name)} characters long")
print(f"position = {position:,.2f}")
```

## 2. A list holds a series of values


```python
returns = [0.012, -0.004, 0.031, 0.008]

print(returns)
print(returns[0], returns[-1])
print(returns[1:3])
print(len(returns))
```

Slices leave out the end index: `returns[1:3]` is elements 1 and 2. A third slot is the
step, so a step of `-1` walks backwards.


```python
word = "level"
print(word[::-1])
print(word == word[::-1])
```

## 3. A loop builds a total


```python
returns = [0.012, -0.004, 0.031, 0.008]

total = 0
count = 0
for r in returns:
    total += r
    count += 1

print(total)
print(count)
print(total / count)
```

`+=` means "add to what is already there". Keeping a total and a count side by side is
the pattern behind almost every function you will write this term.


## 4. Branching


```python
for r in [0.05, -0.02, 0.0]:
    if r > 0:
        label = "gain"
    elif r < 0:
        label = "loss"
    else:
        label = "flat"
    print(r, label)
```
