## Lecture 02 — Distributions: loops, transformations and functions

Returns are not normal — they are wide, they have fat tails, and they have both signs.
The first tool is a loop that walks a list and builds a new one, and the second is
turning that loop into a function so the same transformation can be applied twice.


### The pattern

```python
import math

pos = [2.718282, 4, 55, 31, 7, 81, 1]
new = []
for x in pos:
    new.append(math.log(x))
print(new)
```

Read it as: *start with an empty list, walk the input, append one transformed value per
input value, print the result.*


````python
import math

values = [2.2, -1.6, 4.6, -7.7, 1.1]
total = 0
for value in values:
    total = total + abs(value)
print("Sum of the absolute values:", round(total, 2))
````
