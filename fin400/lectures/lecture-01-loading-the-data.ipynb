## Lecture 01 — Loading the data

The class data comes as plain csv files in `fin400/data/`. Every lecture starts the same
way: read the file, keep the dates as dates, then check the shape before trusting a
single number.

`ticker`, `prc`, `ret` and `mcap` (all money in millions).

