## Lecture 06 — Empirical probability and the shape of a strategy

"How often does this strategy lose more than 5% in a month?" is not a question about a
formula. It is a question about how many months did that — an empirical probability.

### The pattern

```python
ew = data.groupby("caldt")["ret"].mean()          # one return per month
bins = pd.cut(ew * 100, [-100, -5, 0, 5, 100])    # buckets, in percent
share = bins.value_counts().sort_index() / len(ew) * 100
```


````python
import pandas as pd

data = pd.read_csv("fin400-monthly-stocks.csv", parse_dates=["caldt"])
ew = data.groupby("caldt")["ret"].mean()
print("months:", len(ew))
print("mean, percent:", round(float(ew.mean()) * 100, 2))
````
