## Lecture 03 — Statistics with the `statistics` library

The library only formalises what you can already write with a loop. That matters for
one reason: if you cannot write the loop, you do not know what the library is doing.

### The bug worth remembering

```python
maxn = 0
for x in values:
    if x > maxn:
        maxn = x
```

On `[1, 3, 20, 10, 0.5, 2]` this gives 20. On `[-1, -3, -20]` it gives **0**, which is not
even in the list. The fix is to prime with `values[0]`.


````python
import statistics as s

commute = [21, 23, 24, 25, 22]
print("Loop mean:", sum(commute) / len(commute))
print("Library mean:", s.mean(commute))
print("Median:", s.median(commute))
````
