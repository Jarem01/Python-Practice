## Lecture 07 — A first look at pandas

A DataFrame is a table with column names. Three moves cover most of the work: pick a
column, make a new column from old ones, and summarise the result.

### The pattern

```python
df["eps"] = df["ni"] / df["shr"]          # net income per share
df[["ni", "eps"]].describe().round(2)     # one summary for several columns
df["eps"].quantile([0.2, 0.8])             # a percentile, in decimal
```
