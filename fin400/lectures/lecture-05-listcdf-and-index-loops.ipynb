## Lecture 05 — A list of CDF values, and looping by index

The homework pattern: given a list of values and a mean and standard deviation, return
the probability of each one. One function, one loop, one `cdf` call per value.

### The pattern

```python
import statistics as s

def listcdf(data, mu=0, sigma=1):
    nd = s.NormalDist(mu=mu, sigma=sigma)
    out = []
    for x in data:
        out.append(nd.cdf(x))
    return out
```

Then loop over positions to print each value next to its probability — the shape of a
table you would build by hand.


````python
import statistics as s

values = [21.0, 23.0, 25.0]
nd = s.NormalDist(mu=23, sigma=3.9)
for x in values:
    print(x, round(nd.cdf(x), 4))
````
