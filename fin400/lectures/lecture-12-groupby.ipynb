## Lecture 12 — groupby: one summary per group

groupby is the pandas version of "for each group, compute…". Split, apply, combine.

### The pattern

```python
df.groupby('gender')['bmi'].mean()                 # one number per group
df.groupby('gender')[['age', 'bmi']].median()      # several columns
df.groupby(['gender', 'hincome'])['bmi'].median()  # two keys
```


````python
import pandas as pd

df = pd.read_csv("fin400-survey.csv")
print(df["gender"].value_counts())
df.groupby("gender")["age"].mean()
````
