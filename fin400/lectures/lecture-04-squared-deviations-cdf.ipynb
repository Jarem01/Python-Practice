## Lecture 04 — Squared deviations, the CDF and the inverse CDF

Two ideas that the rest of the course leans on: how far values sit from their mean, and
how a normal distribution converts between a value and a probability.

### The pattern

```python
import statistics as s

nd = s.NormalDist(mu=23, sigma=3.9)
nd.cdf(19)            # probability of a value at or below 19
1 - nd.cdf(32)        # probability of a value above 32
nd.cdf(32) - nd.cdf(19)   # between the two
nd.inv_cdf(0.10)      # the value with 10% of the distribution below it
```


````python
import statistics as s

nd = s.NormalDist(mu=23, sigma=3.9)
print(f"x < 19 -> {nd.cdf(19):7.4f}")
print(f"x > 32 -> {1 - nd.cdf(32):7.4f}")
````
