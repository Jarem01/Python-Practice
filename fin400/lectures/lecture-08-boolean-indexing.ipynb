## Lecture 08 — Conditional logic and boolean indexing

Filtering is two steps that look like one: build a True/False column, then use it to pick
rows. Combination is where it usually goes wrong.

### The pattern

```python
mask = (df["gender"] == "male") & (df["age"] >= 40)
older_men = df[mask]
```

Parenthesise every comparison, and use `&` — not `and`.


````python
import pandas as pd

df = pd.read_csv("fin400-survey.csv")
print(df["gender"].unique())
print("aged 40+:", int((df["age"] >= 40).sum()))
````
