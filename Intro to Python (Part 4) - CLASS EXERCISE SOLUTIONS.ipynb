1. Create a function called `count_word` that takes two input parameters: `sentence` and `word`. The `word` parameter is optional with a default value of `'class'`. The function should do the following: 

    1. Convert the `sentence` to lower case.
    2. Remove the following punctuation from the `sentence`: `.`, `!`, `,`, `?` (**HINT**: use the `replace` method).
    3. Create a list containing the words from the `sentence` (**HINT**: use the `split` method).
    4. Create a variable called `count1` equal to the number of times the `word` is included in the list you created in Part 1.3 (**HINT**: use the `count` method).
    5. Return `count1`.

```python
# 1
def count_word(string_var, word='class'):
    string_var = string_var.lower() # Convert the string to lower case to count 'Class' as well as 'class'.
    string_var = string_var.replace('.','') # Remove periods
    string_var = string_var.replace('!','') # Remove exclamation points
    string_var = string_var.replace(',','') # Remove commas
    string_var = string_var.replace('?','') # Remove question marks
    word_list = string_var.split(' ') # Creates a list of words in the string
    count1 = word_list.count(word) # Counts the number of words equal to word (e.g., 'class')
    return count1
```

2. Iterate through the `sentences` list and use the `count_word` function you created in Part 1 to print the following:

            The count of 'class' is {class_count}, and the count of 'semester' is {semester_count}.'

    Fill in the counts of the words `'class'` and `'semester'` for each sentence.

```python
# 2
sentences = ['Wow, I am learning so much in this class this semester!','I wish all of my classes were this awesome.','If I could recommend one class this semester, the data analytics class would be it.','Of all the classes at BYU, this class is the hardest class I have taken and the most rewarding class too.','This class is classy.','I love the Analytics Class this semester!','I love this class.']
for sentence in sentences:
    class_count = count_word(sentence)
    semester_count = count_word(sentence, word='semester')
    print(f"The count of 'class' is {class_count}, and the count of 'semester' is {semester_count}.")
```
