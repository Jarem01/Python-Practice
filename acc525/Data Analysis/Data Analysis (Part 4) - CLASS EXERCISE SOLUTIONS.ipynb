2. You are provided with two DataFrames `data` and `data2`. Merge them based on their common columns to create a new DataFrame called `data`.

```python
data1 = pd.DataFrame({"EmployeeID": [101, 101, 102, 102, 103, 103, 104, 104],"Quarter": ["Q1", "Q2", "Q1", "Q2", "Q1", "Q2", "Q1", "Q2"],"BaseSalary": [15000, 15000, 18000, 18000, 17000, 17000, 16000, 16000],"Department": ["HR", "HR", "IT", "IT", "Finance", "Finance", "Marketing", "Marketing"]})
data2 = pd.DataFrame({"EmployeeID": [101, 101, 102, 103, 103, 105, 105],"Quarter": ["Q1", "Q2", "Q1", "Q1", "Q2", "Q1", "Q2"],"PerformanceScore": [4.5, 4.7, 3.9, 4.2, 4.3, 4.8, 4.6],"Bonus": [2000, 2500, 1500, 1800, 1900, 3000, 3200]})
```

```python
# 2
data = pd.merge(data1, data2, on=['EmployeeID','Quarter'])
data
```

3. Export the `data` DataFrame to a txt file called `mydata.txt` with a `|` as the delimiter. Exclude the index.

```python
# 3
data.to_csv('mydata.txt', sep='|', index=None)
```

#### Exercise 2

You have been provided with 15 csv files in the `data` folder that contain salary data listed by undergraduate major obtained from https://www.kaggle.com/wsj/college-salaries/version/1. Each csv file name contains a letter of the alphabet that makes each file unique. For example, the `'degrees-that-pay-back-a.csv'` file contains all undergraduate majors that start with the letter 'A'

1. Create a variable called `mypath` that points to the `data` folder (this folder was included in today's zipped materials). Then create a list called `filenames` containing all files in the `data` folder.

```python
# 1
from analyticskit import to_excel

mypath = './'  # the data files sit next to this notebook
filenames = os.listdir(mypath)
print(filenames)
```

2. Using the `filenames` list, create a new pandas DataFrame called `df` with all `'degrees-that-pay-back'` csv files concatenated together. Print off the first 10 rows of the concatenated DataFrame to view the result.

```python
# 2
data_list = []
for filename in filenames:
    if 'degrees-that-pay-back' in filename:
        data = pd.read_csv(mypath + filename)
        data_list.append(data)

df = pd.concat(data_list, ignore_index=True)
df.head()
```

3. Export the `df` DataFrame to an Excel file called `mydata.xlsx` into the `data` folder using the `mypath` variable. Exclude the index.

```python
# pandas' to_excel() needs openpyxl or xlsxwriter and the browser has
# neither, so analyticskit writes the .xlsx file itself.
to_excel(df, mypath + 'mydata.xlsx')
```

#### Exercise 3

You have been provided with three csv files. The `earnings.csv` file contains data on earnings and total assets for several companies from 2016 to 2017. The `exchange.csv` file contains the stock exchange on which each company's stock is traded. The `company_names.csv` file contains the name of each company.

Run the cell below to import and view this data:

```python
path = './'  # the data files sit next to this notebook

earnings = pd.read_csv(path + 'earnings.csv',parse_dates=['fpe'])
exchange = pd.read_csv(path + 'exchange.csv')
company = pd.read_csv(path + 'company_names.csv')

print('EARNINGS DATAFRAME:')
display(earnings.head())
print('EXCHANGE DATAFRAME:')
display(exchange.head())
print('COMPANY DATAFRAME:')
display(company.head())
```

1. Merge the `earnings` and `exchange` DataFrames using their common `ticker` identifier. Call the resulting DataFrame `data1`. Print the first 10 rows of `data1`.

```python
# 1
data = pd.merge(earnings,exchange,on=['ticker'])
data.head(10)
```

2. Merge the `data` and `company` DataFrames based on the ticker identifier. Call the resulting DataFrame `data`. Note that the ticker in the `company` DataFrame is called `tic`. Print the first 10 rows of `data`.

```python
# 2
company = company.rename(columns={'tic':'ticker'})
data = pd.merge(data,company,on='ticker')
data.head(10)
```

#### Exercise 4

You are provided with the following DataFrames:

`sales`: Contains monthly sales data for a set of products over the past year.

`spend`: Contains monthly marketing spend for each product over the same period.

`product_info`: Contains product information, including the product name, category, and price.

Run the cell below to import this data.

```python
sales = pd.DataFrame({"product_id": [1, 2, 1, 3, 3, 2, 1, 3, 3, 1],"month": ["2025-01", "2025-01", "2025-02", "2025-02", "2025-03", "2025-03", "2025-04", "2025-04", "2025-05", "2025-05"],"sales": [500, 300, 700, 400, 200, 350, 600, 250, 450, 150]})
spend = pd.DataFrame({"product_id": [1, 2, 1, 3, 2, 1, 4],"month": ["2025-01", "2025-01", "2025-02", "2025-02", "2025-03", "2025-04", "2025-04"],"spend": [100, 50, 120, 80, 60, 110, 90]})
product_info = pd.DataFrame({"product_id": [1, 2, 3, 4, 5],"product_name": ["Product A", "Product B", "Product C", "Product D", "Product E"],"category": ["Electronics", "Home", "Electronics", "Toys", "Garden"],"price": [120, 80, 150, 50, 30]})

print('SALES DATAFRAME:')
display(sales.head())
print('SPEND DATAFRAME:')
display(spend.head())
print('PRODUCT_INFO DATAFRAME:')
display(product_info.head())
```

1. Create a DataFrame called `df` by merging the `sales` and `spend` DataFrames on `product_id` and `month`, making sure that all products with sales data are included, even if they have no corresponding spend data. Print the first 10 rows of `df`.

    You notice that some of the values in the `spend` column in `df` are missing. Explain why the type of merge you used resulted in these missing values.

```python
# 1
df = pd.merge(sales, spend, on=['product_id','month'], how='left')
df.head(10)
```

2. Create a DataFrame called `products_no_sales` by merging the `sales` and `product_info` DataFrames on `product_id`. Your `products_no_sales` DataFrame should include products from the `product_info` DataFrame that are not included in the `sales` DataFrame (i.e., products that have had no sales in the past year).

```python
# 2
products_no_sales = pd.merge(sales, product_info, on='product_id', how='right', indicator=True)
products_no_sales = products_no_sales[products_no_sales['_merge']=='right_only']
products_no_sales.head()
```
