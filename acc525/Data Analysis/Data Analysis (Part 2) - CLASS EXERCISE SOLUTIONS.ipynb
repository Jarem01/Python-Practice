#### Exercise 3

Use the following `df` DataFrame for this exercise, which includes employee names, the hours they worked, and the sales they generated:

```python
df = pd.DataFrame({
    'employee': ['Alice', 'Bob', 'Chris', 'Dana', 'Evan'],
    'hours_worked': [42, 38, 50, 45, 30],
    'sales': [12000, 7000, 18000, 9000, 4000]
})

df.head()
```

1. Use `np.where` to create a new column called `overtime` that equals:

    - `1` if `hours_worked` is greater than `40`

    - `0` if `hours_worked` is less than or equal to `40`

```python
# 1

df['overtime'] = np.where(df['hours_worked'] > 40, 1, 0)
df.head()
```

2. Now suppose the company wants to assign a sales performance level based on the following rules:

    - If `sales` is greater than or equal to `15,000`, performance is `'Excellent'`

    - If `sales` is between `8,000` and `14,999`, performance is `'Good'`

    - If `sales` is less than `8,000`, performance is `'Needs Improvement'`

    Create and apply a function to create a new column called `performance` based on these rules.

```python
# 2

def performance_level(sales):
    if sales >= 15000:
        performance = 'Excellent'
    elif sales >= 8000:
        performance = 'Good'
    else:
        performance = 'Needs Improvement'
    return performance

df['performance'] = df['sales'].apply(performance_level)
df
```

#### Exercise 4

The `df` DataFrame below includes employee work data:

```python
df = pd.DataFrame({
    'employee': ['Alex', 'Brianna', 'Carlos', 'Dina', 'Evan'],
    'hours_worked': [38, 42, 40, 50, 35],
    'hourly_rate': [20, 18, 25, 15, 22]
})

df
```

Create a new column called `total_pay` based on these criteria:

   - If `hours_worked` is **40 or less**:
      - `total_pay` = `hours_worked` × `hourly_rate`

   - If `hours_worked` is **more than 40**:
      - First 40 hours paid at the `hourly_rate`
      - Extra hours paid at 1.5 × `hourly_rate`

```python
def calculate_pay(hours_worked, hourly_rate):
    
    if hours_worked <= 40:
        total_pay = hours_worked * hourly_rate
    else:
        overtime_hours = hours_worked - 40
        total_pay = (40 * hourly_rate) + (overtime_hours * hourly_rate * 1.5)
    
    return total_pay

df['total_pay'] = df.apply(lambda row: calculate_pay(row['hours_worked'], row['hourly_rate']), axis=1)

df
```
