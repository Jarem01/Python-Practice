#### Exercise 1

You are provided with the `skills` list below.

If `'Python'` is not in the `skills` list, print `'What a shame! Python is the best!'`.

Otherwise, if the `skills` list contains exactly 5 items and `'Python'` is the last skill in the list, print `'Saved the best for last.'`

For any other case, print `'Very solid list of skills.'`.

```python
skills = ["JavaScript", "HTML", "SQL", "CSS", "Python"]

if 'Python' not in skills:
    print("What a shame! Python is the best!")
elif len(skills) == 5 and skills[-1] == "Python":
    print("Saved the best for last.")
else:
    print("Very solid list of skills.")
```

#### Exercise 3

Use the `jobs` and `salaries` lists to complete this exercise.

```python
jobs = ['physician','lawyer','r&d manager','software development manager','pharmacy manager','software architect','it manager','pharmacist','finance manager','actuary','tax manager']
salaries = [180000,144500,142120,132000,130000,128250,120000,118000,115000,115000,110000]
```

#### Exercise 4

1. Use a `while` loop to compute 5 factorial. Recall, the formula for 5 factorial is as follows:

        5! = 5 x 4 x 3 x 2 x 1 = 120

**NOTE:** This exercise is meant to stretch you!

```python
# 1

factorial = 1
current_number = 5

while current_number > 0:
    factorial = factorial * current_number
    current_number = current_number - 1
    
print(factorial)
```

2. Use a `while True` loop to repeat the exercise (i.e., to compute 5 factorial).

```python
# 2

factorial = 1
current_number = 5

while True:
    factorial = factorial * current_number
    current_number = current_number - 1
    if current_number == 0:
        break
    
print(factorial)
```
