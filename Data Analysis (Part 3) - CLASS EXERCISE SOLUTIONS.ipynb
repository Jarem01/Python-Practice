#### Exercise 1

You have been provided with a dataset called `invoices.xlsx` which includes invoice data for a fictitious company selling computer parts. 

1. Import this data into a `pandas` DataFrame called `data`, then sort `data` by `InvoiceID` in ascending order and then by `Amount` in descending order. Print the first 5 rows.

```python
# 1

data = pd.read_excel('invoices.xlsx')
data = data.sort_values(by=['InvoiceID', 'Amount'], ascending=[True, False])
data.head()
```

2. Iterate through the rows in `data`. If the `Date` of the transaction is in `May` and the `Model` is `'ProStorage 3TB v4.7'`, print the `Amount`.

    **HINT:** To pull the month of a `Date` variable from a `row`, you can use the code `row.Date.month`. This is different than `df.Date.dt.month` which applies to a full column for extracting the month. For a single row value, we use `row.Date.month`.

```python
# 2

for index,row in data.iterrows():
    if row.Date.month == 5 and row.Model == 'ProStorage 3TB v4.7':
        print(row.Amount)
```

3. Using `data`, identify the `PartName` that was most frequently purchased in the dataset?

```python
counts = data.PartName.value_counts().reset_index()
counts

# Mouse is the PartName most frequently purchased
```

1. Print default summary statistics for the `Quantity`, `Price`, `Amount`, and `High Price` columns in `data`. What percentage of observations have `High Price` equal to `1`?

```python
# 1

data[['Quantity','Price','Amount','High Price']].describe().transpose().reset_index()

# Percentage of Observations with High Price Equal to 1: 18%
```

```python
# 2

data.Amount.describe(percentiles=[.01,.05,.1,.25,.5,.75,.9,.95,.99]).reset_index()

# The maximum value is $1,299,950. The median is $1,949.87. There are clearly some outliers in the distribution!
```

Run the cell below to re-import the `invoices.xlsx` file into the `data` DataFrame variable.

```python
data = pd.read_excel('invoices.xlsx')
data.head()
```

1. Print the summary statistics for the `Amount` column in the `data` DataFrame by `month`. Do you observe anything troubling in the max `Amount` during any month?

```python
# 1

data['month'] = data.Date.dt.month
amounttrend = data.groupby(['month']).Amount.describe().reset_index()
amounttrend

# The max Amount jumps way high in July.
```

2. Based on the answer to #3.1 above, you suspect there may be some suspicious transactions in the month of July. Create a new DataFrame called `july` with observations from `data` that are in the month of July and that have an `Amount` greater than or equal to $10,000. Then use the `july` DataFrame to create a DataFrame called `july_totals` that contains the sum of the `Quantity` and `Amount` columns by `InvoiceID` and `PartName`. What questions might you ask the sales team after viewing this data?

```python
# 2

july = data[(data.month == 7) & (data.Amount > 10000)]
july_totals = july.groupby(['InvoiceID','PartName'])[['Quantity','Amount']].sum().reset_index()
july_totals

# I might ask if this invoice (T13839563) is legitimate. Most quantities purchased are lower than 100. Why did this customer purchase 5000 to 10000 of each type of product?
```
