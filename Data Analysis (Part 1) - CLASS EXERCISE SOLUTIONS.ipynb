2. Create a new Pandas DataFrame called `public12` that includes all rows in our `data` DataFrame where `ownership structure` is equal to 'publicly traded' and the month of the `date` is equal to December. How many observations in our DataFrame are for publicly-traded firms with a date in December?

```python
# 2
public12 = data[(data['ownership structure']=='publicly traded') & (data.date.dt.month == 12)]
public12.info()
public12.head()

# 3,027 observations are publicly traded with a date in December
```
