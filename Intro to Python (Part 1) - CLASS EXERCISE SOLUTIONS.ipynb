3. Calculate the area of a circle with radius 15 following these steps:

    Create a variable called `pi` and assign it a value of 3.14159. 

    Create a variable called `r` and assign it a value of 15. 
    
    Calculate the area of the circle and assign the value to the variable `area`.

```python
# 3
pi = 3.14159
r = 15
area = pi * r**2
```

```python
# 4a
print("Area of a circle with radius " + str(r) + " is equal to " + str(area) + ".")
```
